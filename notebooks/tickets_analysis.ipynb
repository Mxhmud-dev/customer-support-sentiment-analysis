## Sentiment Analysis
Analyzing the sentiment (positive, negative, neutral) of each ticket using TextBlob.

```python
blob = TextBlob(sample_ticket)
print("\nSentiment score:",  blob.sentiment.polarity)
```

```python
def analyze_sentiment(text):
    blob = TextBlob(str(text))
    score = blob.sentiment.polarity

    if score > 0:
        return 'Positive'
    elif score < 0:
        return 'Negative'
    else:
        return 'Neutral'

df['Sentiment'] = df['Ticket Description'].apply(analyze_sentiment)
```

```python
df['Sentiment'].value_counts()
```

## Visualizations
Creating charts to visualize the sentiment distribution.

```python
sentiment_counts = df['Sentiment'].value_counts()

plt.figure(figsize=(8,6))
sentiment_counts.plot(kind='bar', color=['green', 'red', 'gray'])
plt.title('Customer Sentiment Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Number of Tickets')
plt.xticks(rotation=0)
plt.show()
```

```python
sentiment_counts = df['Sentiment'].value_counts()

plt.figure(figsize=(8,6))
sentiment_counts.plot(kind='bar', color=['green', 'red', 'gray'])
plt.title('Customer Sentiment Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Number of Tickets')
plt.xticks(rotation=0)
plt.savefig('../results/sentiment_bar_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(8,6))
colors = ['lightgreen', 'lightcoral', 'lightyellow']
plt.pie(sentiment_counts, labels=sentiment_counts.index, autopct='%1.1f%%', colors=colors, startangle=90)
plt.title('Customer Sentiment Distribution')
plt.savefig('../results/sentiment_pie_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Product Analysis
Identifying which products receive the most negative feedback.

```python
print("=== POSITIVE TICKET EXAMPLE === ")
print(df[df['Sentiment'] == 'Positive']['Ticket Description'].iloc[0])

print("\n=== NEGATIVE TICKET EXAMPLE === ")
print(df[df['Sentiment'] == 'Negative']['Ticket Description'].iloc[0])

print("\n=== NEUTRAL TICKET EXAMPLE === ")
print(df[df['Sentiment'] == 'Neutral']['Ticket Description'].iloc[0])
```

```python
total_tickets = len(df)
positive_count = len(df[df['Sentiment'] == 'Positive'])
negative_count = len(df[df['Sentiment'] == 'Negative'])
neutral_count = len(df[df['Sentiment'] == 'Neutral'])

positive_percent = (positive_count / total_tickets) * 100
negative_percent = (negative_count / total_tickets) * 100
neutral_percent = (neutral_count / total_tickets) * 100

print(f"Total Tickets Analyzed: {total_tickets}")
print(f"\nPositive Tickets: {positive_count} ({positive_percent:.1f}%)")
print(f"Negative: {negative_count} ({negative_percent:.1f}%)")
print(f"Neutral: {neutral_count} ({neutral_percent:.1f}%)")
```

```python
summary_text = f"""Customer Support Ticket Sentiment Analysis Report
================================================

Total Tickets Analyzed: {total_tickets}

Sentiment Breakdown:
-------------------
Positive: {positive_count} ({positive_percent:.1f}%)
Negative: {negative_count} ({negative_percent:.1f}%)
Neutral: {neutral_count} ({neutral_percent:.1f}%)

Key Findings:
-------------
- More than half of the tickets (52%) express positive sentiment
- About 29% of the tickets show negative sentiment
- The remainng 19% are neutral in tone
"""

with open('../results/analysis_summary.text', 'w') as f:
    f.write(summary_text)

print("Summary saved to results/analysis_summary.txt")    
```
